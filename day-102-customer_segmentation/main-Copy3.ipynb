```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
df = pd.read_csv('marketing_campaign.csv', sep='\t')
df.head()
```

```python
df.isna().sum()
```

```python
df = df.dropna()
```

```python
df.duplicated(subset='ID').sum()
```

```python
df.info()
```

```python
data = df.drop(columns='ID')
data.head()
```

```python
data['Dt_Customer'] = pd.to_datetime(data['Dt_Customer'], format = '%d-%m-%Y')
data.info()
```

```python
#add ages
ages = []
for year in data['Year_Birth']:
    ages.append(2025 - year)

data['Age'] = ages
data = data.drop(columns='Year_Birth')
```

```python
data.head()
```

```python
#simplify marital status
status = {"Married":"Partner", "Together":"Partner", "Absurd":"Alone", "Widow":"Alone", "YOLO":"Alone", "Divorced":"Alone", "Single":"Alone"}
data["Living_With"]=data["Marital_Status"].replace(status)
data = data.drop(columns='Marital_Status')
```

```python
data.head()
```

```python
#add children
data["Children"]=data["Kidhome"]+data["Teenhome"]

#Family size
data["family_size"] = data["Living_With"].replace({"Alone": 1, "Partner":2})+ data["Children"]

#Parent True or False
data["is_Parent"] = np.where(data.Children> 0, 1, 0)

data = data.drop(columns=['Kidhome', 'Teenhome'])
data.head()
```

```python
#Segmenting education levels
data["Education"]=data["Education"].replace({"Basic":0,"2n Cycle":0, "Graduation":1, "Master":2, "PhD":2})
data.head()
```

```python
#sum all product
data["Total_MntSpent"] = data["MntWines"]+ data["MntFruits"]+ data["MntMeatProducts"]
+ data["MntFishProducts"]+ data["MntSweetProducts"]+ data["MntGoldProds"]

data = data.drop(columns=['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds'])
data.head()
```

```python
for i in data.columns:
    print(f'column {i}:{data[i].unique()[:5]}')
```

```python
data = data.drop(columns=['Z_Revenue', 'Z_CostContact'])
```

```python
#number of days customer started to shop
newest_cust = data['Dt_Customer'].max()
days = []
for day in data['Dt_Customer']:
    diff = newest_cust - day
    days.append(diff)

data['customer_for'] = days
data['customer_for'] = data['customer_for'].dt.days
data = data.drop(columns='Dt_Customer')
data.head()
```

```python
data['TotalPurchases'] = data['NumDealsPurchases'] + data['NumWebPurchases'] + data['NumCatalogPurchases'] + data['NumStorePurchases']
data = data.drop(columns=['NumDealsPurchases', 'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases'])
```

```python
data['AcceptedCmp1to5'] = data['AcceptedCmp1'] + data['AcceptedCmp2'] + data['AcceptedCmp3'] + data['AcceptedCmp4'] + data['AcceptedCmp5']
data['AcceptedCmp1to5'] = data['AcceptedCmp1to5'].apply(lambda x: 0 if x == 0 else 1)
data = data.drop(columns=['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5'])
data['AcceptedCmp1to5'].value_counts()
```

```python
data["Living_With"] = data["Living_With"].replace({"Alone": 0, "Partner":1})
data['Income'] = data['Income'].astype(int)
```

```python
numeric = ['Income', 'Recency', 'TotalPurchases', 'Age', 'Total_MntSpent', 'customer_for']

for num in numeric:
    sns.boxplot(data[num])
    plt.show()
```

```python
data = data[(data["Age"]<90)]
data = data[(data["Income"]<600000)]
```

```python
# Baseline model with K-means
# As baseline model, I simplified the purpose of segmentation

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from kneed import KneeLocator
from sklearn.metrics import silhouette_score

# Pilih fitur untuk baseline berdasarkan analisis
baseline_features = data[['Age', 'Income', 'Total_MntSpent']].copy()

# Standarisasi fitur
scaler = StandardScaler()
baseline_features_scaled = scaler.fit_transform(baseline_features)

# Elbow Method
inertia_values = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(baseline_features_scaled)
    inertia_values.append(kmeans.inertia_)
# Plot Elbow Method
plt.figure(figsize=(8, 6))
plt.plot(K, inertia_values, 'bx-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.grid(True)
plt.show()

# Deteksi k optimal menggunakan KneeLocator
knee = KneeLocator(K, inertia_values, curve='convex', direction='decreasing')
k_optimal = knee.elbow
print(f'Optimal number of clusters (k) based on Elbow Method: {k_optimal}')
```

```python
# Terapkan K-Means sebagai baseline
baseline_kmeans = KMeans(n_clusters=3, random_state=42)
baseline_clusters = baseline_kmeans.fit_predict(baseline_features_scaled)

# Tambahkan label cluster ke dataframe
data['Baseline_Cluster'] = baseline_clusters

# Visualisasi (proyeksi ke 2D, misalnya Age vs Income)
plt.figure(figsize=(8, 6))
scatter = plt.scatter(data['Age'], data['Income'], c=data['Baseline_Cluster'], cmap='viridis', alpha=0.8)
plt.colorbar(scatter, label='Baseline Cluster')
plt.xlabel('Age')
plt.ylabel('Income')
plt.title('Baseline Clustering Results')
plt.show()

plt.figure(figsize=(8, 6))
scatter = plt.scatter(data['Age'], data['Total_MntSpent'], c=data['Baseline_Cluster'], cmap='viridis', alpha=0.8)
plt.colorbar(scatter, label='Baseline Cluster')
plt.xlabel('Age')
plt.ylabel('Total_MntSpent')
plt.title('Baseline Clustering Results')
plt.show()

plt.figure(figsize=(8, 6))
scatter = plt.scatter(data['Income'], data['Total_MntSpent'], c=data['Baseline_Cluster'], cmap='viridis', alpha=0.8)
plt.colorbar(scatter, label='Baseline Cluster')
plt.xlabel('Income')
plt.ylabel('Total_MntSpent')
plt.title('Baseline Clustering Results')
plt.show()

# Tampilkan inertia
baseline_inertia = baseline_kmeans.inertia_
print(f'Baseline Inertia: {baseline_inertia:.2f}')
silhouette_pca = silhouette_score(baseline_features_scaled, baseline_clusters)
print(f'Silhouette Score with PCA: {silhouette_pca:.2f}')
```

```python
#K-Means with PCA
from sklearn.decomposition import PCA

# Pilih semua fitur numerik relevan
selected_features = data[['Age', 'Income', 'Total_MntSpent', 'customer_for', 'TotalPurchases', 'Recency']].copy()

# Standarisasi
scaler = StandardScaler()
features_scaled = scaler.fit_transform(selected_features)

# PCA
pca = PCA(n_components=3)
features_pca = pca.fit_transform(features_scaled)
print(f'Explained Variance Ratio: {sum(pca.explained_variance_ratio_):.2f}')

# Elbow Method
inertia_values = []
K = range(2, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(features_pca)
    inertia_values.append(kmeans.inertia_)
plt.plot(K, inertia_values, 'bx-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method with PCA')
plt.show()

# Deteksi k optimal
knee = KneeLocator(K, inertia_values, curve='convex', direction='decreasing')
k_optimal = knee.elbow
print(f'Optimal k with PCA: {k_optimal}')

# Terapkan K-Means dengan k optimal
kmeans_pca = KMeans(n_clusters=k_optimal, random_state=42)
clusters_pca = kmeans_pca.fit_predict(features_pca)
inertia_pca = kmeans_pca.inertia_
silhouette_pca = silhouette_score(features_pca, clusters_pca)
print(f'Inertia with PCA (k={k_optimal}): {inertia_pca:.2f}')
print(f'Silhouette Score with PCA: {silhouette_pca:.2f}')

# Visualisasi
plt.figure(figsize=(8, 6))
scatter = plt.scatter(features_pca[:, 0], features_pca[:, 1], c=clusters_pca, cmap='viridis', alpha=0.8)
plt.colorbar(scatter, label='Cluster')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title(f'K-Means with PCA Clustering (k={k_optimal})')
plt.show()
```

Inertia with PCA (k=6): 2994.28
- Baseline dengan k=3 dan 3 fitur asli memiliki inertia 1680.45. Mengingat semakin banyaknya fitur yang digunakan pada model ini dibandingkan baseline

Silhouette Score with PCA: 0.27
- Silhouette Score mengukur kohesi dalam cluster dan pemisahan antar cluster, dengan rentang -1 hingga 1. Nilai 0.27 termasuk dalam kategori "cukup" (0.25-0.5), tetapi masih jauh dari "baik" (>0.5)
